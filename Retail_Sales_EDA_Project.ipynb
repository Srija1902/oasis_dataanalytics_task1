```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/retail_sales_dataset.csv')

df.head()
df.columns
df.shape
df.info()
df.isnull().sum()
df.describe()
df['Total Amount'].mean()
df['Total Amount'].median()
df['Total Amount'].mode()
df['Total Amount'].std()
df['Date'] = pd.to_datetime(df['Date'])
monthly_sales = df.groupby(df['Date'].dt.month)['Total Amount'].sum()
monthly_sales
plt.figure(figsize=(10,5))

monthly_sales.plot(marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.grid(True)

plt.show()
```

```python
product_sales = df.groupby('Product Category')['Total Amount'].sum()

product_sales
plt.figure(figsize=(8,5))

product_sales.plot(kind='bar')

plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')

plt.show()
```

```python
gender_sales = df.groupby('Gender')['Total Amount'].sum()
gender_sales
gender_sales.plot(kind='pie', autopct='%1.1f%%')

plt.title('Sales Distribution by Gender')

plt.ylabel('')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=10)

plt.title('Age Distribution')

plt.xlabel('Age')

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(8,5))

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')

plt.show()
```
